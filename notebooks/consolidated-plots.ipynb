```python
import matplotlib.pyplot as plt
import numpy as np

companies = ["AMZN", "GOOGL", "BLL", "QCOM"]
metrics = ["MAPE", "RMSPE", "RMDPE"]
models = ["Model 1", "Model 2", "Model 3", "Model 4"]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']

# Accuracy data
accuracy_data = {
    "Model 1": {
        "MAPE": [72.20, 76.68, 76.41, 73.72],
        "RMSPE": [44.16, 56.39, 49.03, 46.96],
        "RMDPE": [34.50, 41.27, 41.82, 31.82]
    },
    "Model 2": {
        "MAPE": [70.67, 69.81, 76.07, 72.06],
        "RMSPE": [43.85, 51.67, 50.03, 45.85],
        "RMDPE": [34.55, 37.88, 42.81, 30.99]
    },
    "Model 3": {
        "MAPE": [70.18, 70.84, 76.13, 73.85],
        "RMSPE": [44.24, 51.65, 49.79, 46.90],
        "RMDPE": [34.26, 38.39, 42.74, 31.77]
    },
    "Model 4": {
        "MAPE": [71.79, 74.23, 76.17, 72.47],
        "RMSPE": [43.28, 54.64, 51.26, 46.83],
        "RMDPE": [34.16, 39.56, 43.61, 31.93]
    }
}

# Settings
bar_width = 0.2
bar_spacing = 0.05  # space between model bars
metric_spacing = bar_width * len(models) + 0.1  # space between metrics
company_spacing = 0.5  # additional space between companies

# Compute positions
positions = []
xtick_labels = []
company_centers = []

pos = 0
for company_idx, company in enumerate(companies):
    start = pos
    for metric in metrics:
        positions.append([pos + i * bar_width for i in range(len(models))])
        xtick_labels.append(metric)
        pos += bar_width * len(models) + bar_spacing
    end = pos - bar_spacing
    center = (start + end - bar_width) / 2  # center of group
    company_centers.append(center)
    pos += company_spacing  # extra space between companies

# Flatten the positions for plotting
x_positions_flat = [p for group in positions for p in group]

fig, ax = plt.subplots(figsize=(16, 8))

# Plot bars
for model_idx, model in enumerate(models):
    heights = []
    for company_idx in range(len(companies)):
        for metric in metrics:
            heights.append(accuracy_data[model][metric][company_idx])
    bar_positions = [group[model_idx] for group in positions]
    ax.bar(bar_positions, heights, bar_width, label=model, color=colors[model_idx])

# Metric x-ticks (MAPE, RMSPE, RMDPE)
metric_tick_positions = [np.mean(group) for group in positions]
ax.set_xticks(metric_tick_positions)
ax.set_xticklabels(xtick_labels, rotation=0)

# Company labels centered under groups
for i, company in enumerate(companies):
    ax.text(company_centers[i], -5, company, ha='center', va='top', fontsize=11)

ax.set_ylabel("Accuracy (%)")
ax.set_title("Hybrid GRU to LSTM Validation Accuracies")
ax.legend(title="Models", loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Stocks and metrics
stocks = ['AMZN', 'GOOGL', 'BLL', 'QCOM']
metrics = ['MAPE', 'RMSPE', 'RMDPE']
models = ['Model 1', 'Model 2', 'Model 3', 'Model 4']
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']  # Colors for each model

# Accuracy data structured as [Stock][Model][Metric]
accuracy_data = {
    'AMZN': [
        [94.69, 92.66, 14.78],  # Model 1
        [94.57, 92.66, 14.78],  # Model 2
        [90.97, 88.99, 14.78],  # Model 3
        [90.43, 89.14, 14.78]   # Model 4
    ],
    'GOOGL': [
        [91.50, 89.10, 16.18],
        [93.95, 91.60, 16.18],
        [91.65, 88.57, 16.18],
        [88.43, 86.09, 16.18]
    ],
    'BLL': [
        [96.65, 94.38, 17.22],
        [92.13, 90.83, 17.22],
        [89.09, 88.12, 17.22],
        [97.23, 94.62, 17.22]
    ],
    'QCOM': [
        [95.87, 93.85, 18.50],
        [89.69, 88.59, 18.50],
        [88.40, 87.25, 18.50],
        [93.85, 92.12, 18.50]
    ]
}

# Plotting
fig, axes = plt.subplots(1, 4, figsize=(20, 6), sharey=True)
fig.suptitle('LSTM+GRU Validation Accuracy (Model 1–4)', fontsize=16)

for idx, stock in enumerate(stocks):
    ax = axes[idx]
    stock_data = np.array(accuracy_data[stock])
    x = np.arange(len(metrics))
    bar_width = 0.2

    for model_idx in range(len(models)):
        ax.bar(x + model_idx * bar_width,
               stock_data[model_idx],
               width=bar_width,
               label=models[model_idx],
               color=colors[model_idx])

    ax.set_title(stock)
    ax.set_xticks(x + 1.5 * bar_width)
    ax.set_xticklabels(metrics)
    ax.set_ylim(80, 100 if stock != 'QCOM' else 95)
    ax.set_xlabel('Metric')
    if idx == 0:
        ax.set_ylabel('Accuracy (%)')
    ax.grid(True, axis='y', linestyle='--', alpha=0.7)
    ax.legend(fontsize=9, loc='lower center')

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
